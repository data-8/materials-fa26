## Tables ##

A table is given a name (such as `nba`) in our programming environment so that we can perform operations on the whole table.

```python
# From Basketball Reference and ESPN
nba = Table.read_table('nba_salaries.csv')
nba
```

Manipulating the table creates a new table, but doesn't change/lose the original.

```python
nba.select('name')
```

```python
nba
```

```python
'nba'
```

```python
nba.drop('rank', 'position')
```

```python
nba.sort('name')
```

```python
nba.where('name', "Shaquille O'Neal")
```

```python
nba.where('position', 'PG')
```

Because each operation creates a new table, multiple operations can be applied.

```python
nba.where('position', 'PG').drop('rank', 'position')
```

### **SUMMARIZE**: Constructing a new version of table `t`…

What does each table method do, in "plain English"?

| Num | Table method | Example with `nba` table |
| --- | --- | --- | 
| 1 | `t.select(label)` | `nba.select('position')` |
| 2 | `t.drop(label)` | `nba.drop('position')` |
| 3 | `t.sort(label)` |  `nba.sort('position')` | 
| 4 | `t.where(label, condition)` | `nba.where('position', 'PG')` |

<br/><br/><br/><br/>



The result of applying operations is a table, which can also be given a name (e.g., `point_guards`) in our programming environment.

```python
point_guards = nba.where('position', 'PG').drop('rank', 'position')
point_guards
```

After running the below cell, in what "order" are the rows in the resulting table sorted?

```python
point_guards.where('season', 2026).sort('name').sort('team')
```

Sorting can be performed in descending order as well.

```python
point_guards.where('season', 2026).sort('salary').show(10)
```

Who was the highest-paid Point Guard in 2026?

```python
point_guards.where('season', 2026).sort('salary', descending=True).show(10)
```
